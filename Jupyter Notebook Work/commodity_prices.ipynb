# Import Libraries

```python
import random
import pandas as pd
from datetime import datetime, timedelta
```

# Configuration

```python
NUM_ROWS = 10000

containers = [
    f"C{i:03d}" for i in range(1, 51)
]

products = {
    "P001": "Avocado",
    "P002": "Mango",
    "P003": "Tomato",
    "P004": "Apple"
}

markets = [
    "Delhi",
    "Jaipur",
    "Mumbai",
    "Pune",
    "Ahmedabad"
]

start_date = datetime(2026, 1, 1)
end_date = datetime(2026, 6, 30)
```

# Random Date Function

```python
def random_date():

    days = random.randint(
        0,
        (end_date - start_date).days
    )

    return start_date + timedelta(days=days)
```

# Generate Commodity Price Data

```python
data = []

for i in range(NUM_ROWS):

    # Select container
    container_id = random.choice(containers)

    # Select product
    product_id = random.choice(list(products.keys()))

    # Product name
    commodity = products[product_id]

    # Select market
    market = random.choice(markets)

    # Commodity-specific price
    if product_id == "P001":       # Avocado
        price = random.uniform(150, 350)

    elif product_id == "P002":     # Mango
        price = random.uniform(80, 250)

    elif product_id == "P003":     # Tomato
        price = random.uniform(30, 120)

    else:                           # Apple
        price = random.uniform(100, 300)

    record = {
        "container_id": container_id,
        "product_id": product_id,
        "commodity": commodity,
        "market": market,
        "price": round(price, 2),
        "date": random_date()
    }

    data.append(record)
```

# Create DataFrame

```python
commodity_df = pd.DataFrame(data)
```

# Check Data

```python
print("Data generation completed!")
print("Total Rows:", len(commodity_df))
print("Total Columns:", len(commodity_df.columns))
print("Shape:", commodity_df.shape)

display(commodity_df.head(10))
```
