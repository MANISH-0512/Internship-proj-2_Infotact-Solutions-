# Import Kafka

```python
import json
import random
import time

from datetime import datetime
from kafka import KafkaProducer
```

# Kafka Producer connect

```python
producer = KafkaProducer(
    bootstrap_servers="localhost:9092",
    value_serializer=lambda v: json.dumps(v).encode("utf-8")
)

print("Kafka Producer connected successfully!")
```

# IoT event generator

```python
containers = [f"C{i:03d}" for i in range(1, 51)]

products = {
    "P001": "Avocado",
    "P002": "Mango",
    "P003": "Tomato",
    "P004": "Apple"
}

locations = [
    "Delhi",
    "Jaipur",
    "Mumbai",
    "Pune",
    "Ahmedabad"
]


def generate_iot_event():

    product_id = random.choice(list(products.keys()))

    event = {
        "event_id": f"E{random.randint(100000, 999999)}",

        "container_id": random.choice(containers),

        "product_id": product_id,

        "commodity": products[product_id],

        "timestamp": datetime.now().isoformat(),

        "temperature": round(
            random.uniform(4, 16), 2
        ),

        "humidity": round(
            random.uniform(60, 95), 2
        ),

        "vibration": round(
            random.uniform(0, 1), 2
        ),

        "latitude": round(
            random.uniform(20, 30), 6
        ),

        "longitude": round(
            random.uniform(70, 80), 6
        ),

        "location": random.choice(locations)
    }

    return event
```

# Send a first message to kafka

```python
event = generate_iot_event()

producer.send(
    "container_telemetry",
    value=event
)

producer.flush()

print("Event sent to Kafka:")
print(event)
```

# Continuous streaming

```python
print("AtmoSync IoT Kafka Producer Started...")
print("Press Stop/Interrupt to stop.\n")

try:

    while True:

        event = generate_iot_event()

        producer.send(
            "container_telemetry",
            value=event
        )

        producer.flush()

        print(
            f"Sent: {event['container_id']} | "
            f"{event['commodity']} | "
            f"Temp: {event['temperature']}°C | "
            f"Humidity: {event['humidity']}%"
        )

        time.sleep(2)

except KeyboardInterrupt:

    print("\nProducer stopped.")

finally:

    producer.close()
```
