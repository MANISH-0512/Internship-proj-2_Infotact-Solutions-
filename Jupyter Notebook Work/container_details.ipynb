# Import libraries

```python
import random
import pandas as pd
from datetime import datetime, timedelta
```

# Configuration

```python
NUM_ROWS = 10000

containers = [
    f"C{i:03d}" for i in range(1, 51)
]

products = [
    "P001",
    "P002",
    "P003",
    "P004"
]

locations = [
    "Delhi",
    "Jaipur",
    "Mumbai",
    "Pune",
    "Ahmedabad"
]

start_date = datetime(2026, 1, 1)
end_date = datetime(2026, 6, 30)
```

# Random Dispatch Date

```python
def random_date():

    days = random.randint(
        0,
        (end_date - start_date).days
    )

    return start_date + timedelta(days=days)
```

# Generate Container Data

```python
data = []

for i in range(NUM_ROWS):

    source = random.choice(locations)

    destination = random.choice(locations)

   
    while destination == source:
        destination = random.choice(locations)

    record = {

        "container_id": random.choice(containers),

        # Common column
        "product_id": random.choice(products),

        "quantity": random.randint(
            100, 2000
        ),

        "source": source,

        "destination": destination,

        "dispatch_date": random_date()
    }

    data.append(record)
```

# Create DataFrame

```python
container_df = pd.DataFrame(data)
```

# Check Data

```python
print("Data generation completed!")
print("Total Rows:", len(container_df))
print("Total Columns:", len(container_df.columns))
print("Shape:", container_df.shape)

display(container_df.head(10))
```
