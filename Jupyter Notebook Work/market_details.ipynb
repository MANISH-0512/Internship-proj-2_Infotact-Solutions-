# Import Libraries

```python
import random
import pandas as pd
```

# Configuration

```python
NUM_ROWS = 10000

products = [
    "P001",
    "P002",
    "P003",
    "P004"
]


markets = [
    {
        "market_id": "M001",
        "market": "Delhi Market",
        "city": "Delhi",
        "state": "Delhi"
    },
    {
        "market_id": "M002",
        "market": "Jaipur Market",
        "city": "Jaipur",
        "state": "Rajasthan"
    },
    {
        "market_id": "M003",
        "market": "Mumbai Market",
        "city": "Mumbai",
        "state": "Maharashtra"
    },
    {
        "market_id": "M004",
        "market": "Pune Market",
        "city": "Pune",
        "state": "Maharashtra"
    },
    {
        "market_id": "M005",
        "market": "Ahmedabad Market",
        "city": "Ahmedabad",
        "state": "Gujarat"
    }
]
```

# Generate Market Details

```python
data = []

for i in range(NUM_ROWS):

    # Random market
    market_info = random.choice(markets)

    # Random product
    product_id = random.choice(products)

    record = {

        "market_id": market_info["market_id"],

        # Common product key
        "product_id": product_id,

        "market": market_info["market"],

        "city": market_info["city"],

        "state": market_info["state"]
    }

    data.append(record)
```

# Create DataFrame

```python
market_df = pd.DataFrame(data)
```

# Check Data

```python
print("Data generation completed!")
print("Total Rows:", len(market_df))
print("Total Columns:", len(market_df.columns))
print("Shape:", market_df.shape)

display(market_df.head(10))
```
