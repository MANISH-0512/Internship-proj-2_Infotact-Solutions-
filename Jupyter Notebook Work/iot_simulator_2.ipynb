# Step - 1 :- IMPORT LIBRARIES

```python
import random
import time
from datetime import datetime
```

# Step - 2 :- CONFIGURATION

```python

containers = [f"C{i:03d}" for i in range(1, 51)]


products = {
    "P001": "Avocado",
    "P002": "Mango",
    "P003": "Tomato",
    "P004": "Apple"
}


locations = [
    "Delhi",
    "Jaipur",
    "Mumbai",
    "Pune",
    "Ahmedabad"
]

print("IoT Simulator Configuration Ready!")
print("Containers:", len(containers))
print("Products:", len(products))
```

# Step - 3 :- Generate One IoT Event

```python
def generate_iot_event():

    product_id = random.choice(list(products.keys()))

    event = {
        "event_id": f"E{random.randint(100000, 999999)}",

        "container_id": random.choice(containers),

        "product_id": product_id,

        "commodity": products[product_id],

        "timestamp": datetime.now().isoformat(),

        "temperature": round(
            random.uniform(4, 16), 2
        ),

        "humidity": round(
            random.uniform(60, 95), 2
        ),

        "vibration": round(
            random.uniform(0, 1), 2
        ),

        "latitude": round(
            random.uniform(20, 30), 6
        ),

        "longitude": round(
            random.uniform(70, 80), 6
        ),

        "location": random.choice(locations)
    }

    return event
```

# Step - 4 :- TEST EVENT

```python
event = generate_iot_event()

print(event)
```

# Step - 5 :- CONTINOUS IOT SIMULATION

```python
print("AtmoSync IoT Simulator Started...")
print("Press Interrupt/Stop to stop the simulator.\n")

while True:

    event = generate_iot_event()

    print(event)

    # New sensor event every 2 seconds
    time.sleep(2)
```
