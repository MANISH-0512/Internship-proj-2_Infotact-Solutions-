# Import Libraries

```python
import random
import pandas as pd
from datetime import datetime, timedelta
```

# Configuration

```python
NUM_ROWS = 10000

containers = [f"C{i:03d}" for i in range(1, 51)]

products = {
    "P001": "Avocado",
    "P002": "Mango",
    "P003": "Tomato",
    "P004": "Apple"
}

locations = [
    "Delhi",
    "Jaipur",
    "Mumbai",
    "Pune",
    "Ahmedabad"
]

start_date = datetime(2026, 1, 1)
end_date = datetime(2026, 6, 30)
```

# Random Timestamp Function

```python
def random_datetime():

    days = random.randint(
        0,
        (end_date - start_date).days
    )

    date = start_date + timedelta(days=days)

    hour = random.randint(0, 23)
    minute = random.randint(0, 59)
    second = random.randint(0, 59)

    return date.replace(
        hour=hour,
        minute=minute,
        second=second
    )
```

# Generate IoT Sensor Data

```python
data = []

for i in range(NUM_ROWS):

    product_id = random.choice(list(products.keys()))

    record = {
        "event_id": f"E{i+1:06d}",

        "container_id": random.choice(containers),

        # Common column
        "product_id": product_id,

        "timestamp": random_datetime(),

        "temperature": round(
            random.uniform(4, 16), 2
        ),

        "humidity": round(
            random.uniform(60, 95), 2
        ),

        "vibration": round(
            random.uniform(0, 1), 2
        ),

        "latitude": round(
            random.uniform(20, 30), 6
        ),

        "longitude": round(
            random.uniform(70, 80), 6
        ),

        "location": random.choice(locations)
    }

    data.append(record)
```

# Create DataFrame

```python
df = pd.DataFrame(data)
```

# Check Data

```python

print("Data generation completed!")
print("Total Rows:", len(df))
print("Total Columns:", len(df.columns))
print("Shape:", df.shape)

display(df.head(10))
```

# Save CSV

```python
df.to_csv(
    r"C:\Users\Dell\Downloads\iot_sensor_data.csv",
    index=False
)

print("\nCSV file created successfully!")
print(r"C:\Users\Dell\Downloads\iot_sensor_data.csv")
```
